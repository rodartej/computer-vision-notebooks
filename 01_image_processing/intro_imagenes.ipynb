```python
import cv2
import matplotlib.pyplot as plt
import numpy as np
```

```python
matrizDeCeros = np.zeros((100,50)) # filas, columnas
matrizDeUnos = np.ones((100,50))

imagen = np.uint8(np.concatenate((matrizDeCeros, matrizDeUnos), axis=1)) # juntar las dos matrices para formar una
plt.imshow(imagen, cmap='gray') # mitad negra - mitad blanca
```

```python
gx = cv2.Sobel(imagen, cv2.CV_64F, 1, 0, ksize=5) # gradiente en x
gy = cv2.Sobel(imagen, cv2.CV_64F, 0, 1, ksize=5) # gradiente en y

magnitud, angulo = cv2.cartToPolar(gx, gy)
```

```python
gx = cv2.convertScaleAbs(gx) # convertir a uint8
gy = cv2.convertScaleAbs(gy)
magnitud = cv2.convertScaleAbs(magnitud)
```

```python
angulo = (180 / np.pi) * angulo # convertir a grados
angulo
```
