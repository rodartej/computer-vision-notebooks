```python
X = [(0,0), (0,1), (1,0), (1,1)]
y = [0, 0, 0, 1]

w1, w2 = -0.3, 0.2
umbral = 0
a = 0.2 # coeficiente de aprendizaje
max_epocas = 48

print(f"Pesos iniciales: w1={w1}, w2={w2} | umbral={umbral} | a={a}\n")

for epoca in range(1, max_epocas+1):
    print(f"===== Epoca {epoca} =====")
    errores = 0
    for i in range(4):
        x1, x2 = X[i]
        esp = y[i]
        suma = w1*x1 + w2*x2
        real = 1 if suma >= umbral else 0
        error = esp - real
        w1_ant, w2_ant = w1, w2
        if error != 0:
            w1 = w1 + a * x1 * error
            w2 = w2 + a * x2 * error
            print(f"  ({x1},{x2}) esp={esp} suma={suma:.2f} real={real} err={error:+d} -> CORRIGE: w1 {w1_ant:.2f}->{w1:.2f}, w2 {w2_ant:.2f}->{w2:.2f}")
            errores += 1
        else:
            print(f"  ({x1},{x2}) esp={esp} suma={suma:.2f} real={real} err=0   -> IGUAL:    w1={w1:.2f}, w2={w2:.2f}")
    if errores == 0:
        print(f"\nConvergio en epoca {epoca}")
        print(f"Pesos finales: w1={w1:.2f}, w2={w2:.2f}")
        break
    print()
```
