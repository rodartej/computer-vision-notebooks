```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
imagen = cv2.imread('bananas.jpg')
imagen = cv2.cvtColor(imagen, cv2.COLOR_BGR2RGB)
plt.imshow(imagen)
plt.show()
```

```python
kernel_3x3 = np.ones((3, 3), dtype=np.float32) / 9
salida = cv2.filter2D(imagen, -1, kernel_3x3)

plt.imshow(salida)
plt.title('imagen borrosa')
plt.show()
```

```python
kernel_5x5 = np.ones((5, 5), dtype=np.float32) / 25
salida_5x5 = cv2.filter2D(imagen, -1, kernel_5x5)

plt.imshow(salida_5x5)
plt.title('imagen borrosa con kernel 5x5')
plt.show()
```

```python
kernel_31x31 = np.ones((31, 31), dtype=np.float32) / 961
salida_31x31 = cv2.filter2D(imagen, -1, kernel_31x31)

plt.imshow(salida_31x31)
plt.title('imagen borrosa con kernel 31x31')
plt.show()
```

```python
# filtro gaussiano
for kernel_size in [5, 11, 21, 31]:
    salida = cv2.GaussianBlur(imagen, (kernel_size, kernel_size), 0)
    plt.imshow(salida)
    plt.title(f'filtro gaussiano {kernel_size}x{kernel_size}')
    plt.show()
```
