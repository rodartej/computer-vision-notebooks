```python
import math
import pandas as pd

# 1. Definimos el punto de entrada P
P = (3, 2)

# 2. Lista de patrones (x, y, clase) extraídos de la imagen
patrones = [
    (0.8, 0.8, 1), (1.0, 1.0, 1), (1.2, 0.8, 1), (0.8, 1.2, 1), (1.2, 1.2, 1),
    (4.0, 3.0, 2), (3.8, 2.8, 2), (4.2, 2.8, 2), (3.8, 3.2, 2), (4.2, 3.2, 2), (4.4, 2.8, 2), (4.4, 3.2, 2),
    (3.2, 0.4, 3), (3.2, 0.7, 3), (3.8, 0.5, 3), (3.5, 1.0, 3), (4.0, 1.0, 3), (4.0, 0.7, 3)
]

distancias_lista = []

# 3. Ciclo for para calcular la distancia a cada punto X_n
print(f"{'Punto':<10} | {'Coordenadas':<15} | {'Clase':<6} | {'Distancia a P':<15}")
print("-" * 55)

for i, punto in enumerate(patrones, 1):
    px, py, clase = punto
    
    # Cálculo de distancia euclidiana
    distancia = math.sqrt((px - P[0])**2 + (py - P[1])**2)
    
    # Guardamos para un análisis posterior
    distancias_lista.append({
        'Punto': f'X{i}',
        'x': px,
        'y': py,
        'clase': clase,
        'distancia': distancia
    })
    
    print(f"X{i:<8} | ({px:>3}, {py:>3})     | {clase:<6} | {distancia:.4f}")

# 4. (Opcional) Encontrar el punto más cercano
df_distancias = pd.DataFrame(distancias_lista)
mas_cercano = df_distancias.loc[df_distancias['distancia'].idxmin()]

print("\n" + "="*55)
print(f"El punto más cercano es {mas_cercano['Punto']} (Clase {mas_cercano['clase']}) "
      f"con una distancia de {mas_cercano['distancia']:.4f}")
```

```python
import numpy as np
from collections import Counter

# 1. Configuración de datos y Punto P
P = np.array([4.2, 1.8])
k = 5

datos = np.array([
    [0.8, 0.8, 1], [1.0, 1.0, 1], [1.2, 0.8, 1], [0.8, 1.2, 1], [1.2, 1.2, 1],
    [4.0, 3.0, 2], [3.8, 2.8, 2], [4.2, 2.8, 2], [3.8, 3.2, 2], [4.2, 3.2, 2], [4.4, 2.8, 2], [4.4, 3.2, 2],
    [3.2, 0.4, 3], [3.2, 0.7, 3], [3.8, 0.5, 3], [3.5, 1.0, 3], [4.0, 1.0, 3], [4.0, 0.7, 3]
])

# 2. Cálculo de Distancias Euclidianas
coords = datos[:, :2]
clases = datos[:, 2]
distancias = np.linalg.norm(coords - P, axis=1)

# 3. Obtener los 5 vecinos más cercanos
indices_k = np.argsort(distancias)[:k]
vecinos_dist = distancias[indices_k]
vecinos_clases = clases[indices_k]

# 4. Cálculo de Pesos Individuales y Acumulados (Mk-NN)
pesos_por_clase = {}
print(f"--- Desglose de Pesos Individuales (k={k}) ---")
print(f"{'Vecino':<10} | {'Clase':<6} | {'Distancia (d)':<15} | {'Peso (1/d)':<15}")
print("-" * 55)

for i in range(k):
    c = int(vecinos_clases[i])
    d = vecinos_dist[i]
    
    # Cálculo del 'valorcito' o peso individual
    peso_individual = 1 / d if d != 0 else 1e10
    
    # Acumular en la clase correspondiente
    pesos_por_clase[c] = pesos_por_clase.get(c, 0) + peso_individual
    
    print(f"Vecino {i+1:<3} | {c:<6} | {d:<15.4f} | {peso_individual:<15.4f}")

# 5. Resultados Finales
print("\n--- Sumatoria Final de Pesos por Clase ---")
for clase, peso_total in pesos_por_clase.items():
    print(f"Clase {clase}: {peso_total:.4f}")

# Decisión final
clase_ganadora = max(pesos_por_clase, key=pesos_por_clase.get)
print(f"\nResultado Mk-NN: El punto se asigna a la Clase {clase_ganadora}")
```

```python
import numpy as np
import pandas as pd
from collections import Counter

# 1. Configuración de datos
P = np.array([4.2, 1.8])
k = 5

# Matriz de entrenamiento: [x, y, clase]
datos = np.array([
    [0.8, 0.8, 1], [1.0, 1.0, 1], [1.2, 0.8, 1], [0.8, 1.2, 1], [1.2, 1.2, 1],
    [4.0, 3.0, 2], [3.8, 2.8, 2], [4.2, 2.8, 2], [3.8, 3.2, 2], [4.2, 3.2, 2], [4.4, 2.8, 2], [4.4, 3.2, 2],
    [3.2, 0.4, 3], [3.2, 0.7, 3], [3.8, 0.5, 3], [3.5, 1.0, 3], [4.0, 1.0, 3], [4.0, 0.7, 3]
])

# 2. Cálculo de Distancias
coords = datos[:, :2]
clases = datos[:, 2]

# Distancia Euclidiana: sqrt((x-px)^2 + (y-py)^2)
distancias = np.linalg.norm(coords - P, axis=1)

# 3. Selección de los k vecinos más cercanos
indices_ordenados = np.argsort(distancias)
k_indices = indices_ordenados[:k]

vecinos_dist = distancias[k_indices]
vecinos_clases = clases[k_indices]

# --- ALGORITMO 1: k-NN (Votación simple) ---
conteo_votos = Counter(vecinos_clases)
clase_knn = conteo_votos.most_common(1)[0][0]

# --- ALGORITMO 2: Mk-NN (Votación ponderada) ---
pesos_por_clase = {}
for c, d in zip(vecinos_clases, vecinos_dist):
    # Evitar división por cero si d=0
    peso = 1 / d if d != 0 else 1e10
    pesos_por_clase[c] = pesos_por_clase.get(c, 0) + peso

clase_mknn = max(pesos_por_clase, key=pesos_por_clase.get)

# 4. Mostrar Resultados
print(f"Análisis para P = {P} con k = {k}")
print("-" * 40)
print(f"Vecinos seleccionados (Clase | Distancia):")
for i in range(k):
    print(f" - Clase {int(vecinos_clases[i])} | Dist: {vecinos_dist[i]:.4f}")

print("\n--- RESULTADOS ---")
print(f"Clasificación k-NN:   Clase {int(clase_knn)} (Votos: {dict(conteo_votos)})")
print(f"Clasificación Mk-NN:  Clase {int(clase_mknn)}")
print("Pesos Mk-NN (1/d):")
for c, p in pesos_por_clase.items():
    print(f" - Clase {int(c)}: {p:.4f}")
```
